```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np
cars_df=pd.read_csv('/content/Cars93 (2).csv')
print(cars_df)
fig,ax=plt.subplots(2,3)
fig.set_figwidth(10)
fig.set_figheight(7)
fig.suptitle('multiple box plots',fontsize=16)
ax[0][0].boxplot(cars_df["Price"][cars_df["Type"]=="Compact"])
ax[0][0].sub_title=('Compact')
ax[0][1].boxplot(cars_df["Price"][cars_df["Type"]=="Midsize"])
ax[0][1].set_title=('Midsize')
ax[0][2].boxplot(cars_df["Price"][cars_df["Type"]=="Large"])
ax[0][2].set_title=('Large')
ax[1][0].boxplot(cars_df["Price"][cars_df["Type"]=="Small"])
ax[1][0].set_title=('Small')
ax[1][1].boxplot(cars_df["Price"][cars_df["Type"]=="Sporty"])
ax[1][1].set_title=('Sporty')
ax[1][2].boxplot(cars_df["Price"][cars_df["Type"]=="Van"])
ax[1][2].set_title=('Van')
plt.hist(cars_df['MPG.city'],color='black')
plt.xlabel('MGP.city')
plt.ylabel('city')
plt.show()

plt.plot(cars_df['EngineSize'],cars_df['Horsepower'])
plt.xlabel('EngineSize')
plt.ylabel('Horsepower')
plt.title('linecharttoobservethevariations"EngineSize"againt"Horsepower"')


```
